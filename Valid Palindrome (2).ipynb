```python
class Solution:

    def isAlphanumeric(self, s):
        x = ord(s)
        if 97 <= x <= 122 or 65 <= x <= 90 or 48 <= x <= 57:
            return True
        return False

    def isPalindrome(self, s: str) -> bool:
        s = s.lower()
        i = 0
        j = len(s) - 1
        while i < j:
            if not self.isAlphanumeric(s[i]):
                i += 1
            elif not self.isAlphanumeric(s[j]):
                j -= 1
            elif s[i] == s[j]:
                i += 1
                j -= 1
            else:
                return False
        return True


# --- Jupyter Notebook Execution ---
# 1. Instantiate the class
solution = Solution()

# 2. Define a test string
test_string = "A man, a plan, a canal: Panama"

# 3. Call the method and store the result
result = solution.isPalindrome(test_string)

# 4. Print the output
print(f"Is '{test_string}' a palindrome? -> {result}")
```
