```python
from typing import List

class Solution:
    def reverseString(self, s: List[str]) -> None:
        """
        Do not return anything, modify s in-place instead.
        """
        i = 0
        j = len(s) - 1

        while i < j:
            temp = s[i]
            s[i] = s[j]
            s[j] = temp
            i += 1
            j -= 1

# --- TEST THE CODE ---
# 1. Create an instance of the Solution class
solver = Solution()

# 2. Define a list of characters (representing a string)
char_list = ["h", "e", "l", "l", "o"]
print("Original:", char_list)

# 3. Call the function (it modifies char_list in-place)
solver.reverseString(char_list)

# 4. Print the result
print("Reversed:", char_list)
```
