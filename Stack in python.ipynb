```python
#List Implementation of Stack

class Stack:

  def __init__(self):
    self.st = []

  def push(self, x):
    self.st.append(x)

  def pop(self):
    if len(self.st) == 0:
      return "-1"
    x = self.st[-1]
    self.st.pop()  # Fixed from st.pop() to self.st.pop()
    return x

  def top(self):
    if len(self.st) == 0:
      return "-1"
    return self.st[-1]  # Added return statement and proper indentation

  def size(self):
    return len(self.st)  # Added colon and moved return to a new indented line

stack = Stack()

stack.push(5)
stack.push(2)
stack.push(1)
stack.push(9)

print(stack.size())
print(stack.pop())
print(stack.top())
```
