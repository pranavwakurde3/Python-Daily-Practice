```python
from typing import Optional

# 1. Define the ListNode class
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

# 2. Paste your Solution class
class Solution:
    def middleNode(self, head: Optional[ListNode]) -> Optional[ListNode]:
        slow = head
        fast = head

        while fast != None and fast.next != None:
            slow = slow.next
            fast = fast.next.next
        
        return slow

# 3. Helper functions to test the code in Jupyter
def create_linked_list(arr):
    """Converts a standard Python list into a Linked List."""
    if not arr:
        return None
    head = ListNode(arr[0])
    current = head
    for val in arr[1:]:
        current.next = ListNode(val)
        current = current.next
    return head

def print_linked_list(head):
    """Prints the linked list sequence visually."""
    elements = []
    current = head
    while current:
        elements.append(str(current.val))
        current = current.next
    print(" -> ".join(elements) if elements else "Empty List")

# --- Test Execution ---
# Create a linked list: 1 -> 2 -> 3 -> 4 -> 5
linked_list = create_linked_list([1, 2, 3, 4, 5])
print("Original List:")
print_linked_list(linked_list)

# Find and print the middle node sequence
sol = Solution()
middle = sol.middleNode(linked_list)
print("\nFrom Middle Node onwards:")
print_linked_list(middle)
```
