```python
class Solution:
    def sortArray(self, nums: list[int]) -> list[int]:
        n = len(nums)
        for i in range(n):
            isSwap = False
            for j in range(n - i - 1):
                if nums[j] > nums[j + 1]:
                    # swap
                    temp = nums[j]
                    nums[j] = nums[j + 1]
                    nums[j + 1] = temp
                    isSwap = True
            if not isSwap:
                break
        return nums

# Test the code in your Jupyter cell
sol = Solution()
sample_nums = [64, 34, 25, 12, 22, 11, 90]
sorted_nums = sol.sortArray(sample_nums)
print("Sorted array:", sorted_nums)
```
