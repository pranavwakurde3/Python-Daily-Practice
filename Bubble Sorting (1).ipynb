```python
class Solution:
    def bubbleSort(self, nums):
        n = len(nums)  # 1. Defined 'n'
        for i in range(n):
            isSwap = False
            for j in range(n - i - 1):
                if nums[j] > nums[j + 1]:
                    # swap (Pythonic way)
                    nums[j], nums[j + 1] = nums[j + 1], nums[j]
                    isSwap = True
            if not isSwap:  # 2. Removed the colon after break
                break
        return nums  # 4. Corrected indentation


if __name__ == "__main__":
    solution = Solution()
    nums = [7, 4, 1, 5, 3]
    print("Array Before Using Bubble Sort:", nums)

    # 3. Changed bubble_sort to bubbleSort to match the class method name
    sorted_nums = solution.bubbleSort(nums)
    print("Array After Using Bubble Sort:", sorted_nums)
```
