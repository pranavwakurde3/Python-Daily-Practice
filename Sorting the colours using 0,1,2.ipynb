```python
class Solution:
    def sortColors(self, nums: list[int]) -> None:
        """
        Do not return anything, modify nums in-place instead.
        """
        left = 0
        right = len(nums) - 1
        i = 0

        while i <= right:
            if nums[i] == 1:
                i += 1
            elif nums[i] == 0:
                # 1. Swap
                temp = nums[i]
                nums[i] = nums[left]
                nums[left] = temp
                # 2. Fix: Move both pointers forward
                left += 1
                i += 1
            else:
                # 1. Swap
                temp = nums[i]
                nums[i] = nums[right]
                nums[right] = temp
                # 2. Fix: Shrink the right boundary (do NOT increment i yet)
                right -= 1

# --- Jupyter Notebook Execution Code ---

# 1. Create a test case
my_list = [2, 0, 2, 1, 1, 0]
print(f"Original list: {my_list}")

# 2. Instantiate the Solution class
solution = Solution()

# 3. Call the method (it modifies the list in-place)
solution.sortColors(my_list)

# 4. Print the sorted result
print(f"Sorted list:   {my_list}")
```
