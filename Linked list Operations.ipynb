```python
class Node:

    def __init__(self, data):
        self.data = data
        self.next = None
```

```python
a = Node(5)
b = Node(3)
c = Node(7)

a.next = b
b.next = c

head = a
```

```python
print(head.data)
print(head.next.data)
print(head.next.next.data)
```

```python
# traversed linked list
def printLinkedList(head):
    curr = head

    while curr != None:
        print(curr.data, end=" ")
        curr = curr.next
        
printLinkedList(head)
```

```python
#Insertion at beginning of LL

newNode = Node(4)
newNode.next = head
head = newNode

printLinkedList(head)
```

```python
#Insertion at end of LL

newNode = Node(1)

curr = head
while curr.next!=None:
    curr = curr.next

curr.next = newNode
printLinkedList(head)
```

```python
# insertion at the kth index

k=2
newNode = Node(6)

curr = head
for i in range(k-1):
  curr = curr.next

newNode.next = curr.next
curr.next = newNode

printLinkedList(head)
```

```python
#Deletion the first node

head = head.next
printLinkedList(head)
```

```python
#Deletion the last node

curr = head
while curr.next.next!=None:
    curr = curr.next

curr.next = None
printLinkedList(head)
```

```python
#delete kth index

k=2

curr = head
for i in range(k-1):
  curr = curr.next

curr.next = curr.next.next
printLinkedList(head)
```
