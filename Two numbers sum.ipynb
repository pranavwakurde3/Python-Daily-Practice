```python
from typing import List

class Solution:
    def twoSum(self, nums: List[int], target: int) -> List[int]:
        n = len(nums)
        dict1 = {}
        for i in range(n):
            rem = target - nums[i]
            if rem in dict1:
                return [dict1[rem], i]
            dict1[nums[i]] = i

# --- Test the code ---
# 1. Create an instance of the class
solution = Solution()

# 2. Define test inputs
nums_input = [2, 7, 11, 15]
target_input = 9

# 3. Call the method and print the result
result = solution.twoSum(nums_input, target_input)
print(f"Indices: {result}")
```
